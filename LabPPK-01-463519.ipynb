### <span style="color:blue">Kerja #5: Kemiringan dan Panjang Garis &#8594; 30 poin</span>

Pada suatu bidang datar, suatu garis lurus dapat didefinisikan mengggunakan koordinat dari titik pangkal dan titik ujungnya. Koordinat titik pangkal dapat dinyatakan dengan $(x_1, y_1)$ dan koordinat titik ujung dapat dinyatakan dengan $(x_2, y_2)$.

Ketika suatu garis lurus sudah terdefinisikan maka dapat ditentukan kemiringan- dan panjang-nya.
- Kemiringan garis dapat ditentukan memakai rumus: $$m = \frac{y_2 - y_1}{x_2 - x_1}$$
- Panjang garis dapat ditentukan memakai rumus: $$l = \sqrt{(x_2 - x_1)^2 + (y_2 - y_1)^2}$$

<span style="color:red">&#9881;</span> Berdasarkan dua rumus di atas, pada sel di bawah ini buatlah program Python lengkap yang menampilkan kemiringan dan panjang dari suatu garis ketika koordinat titik pangkal dan koordinat titik ujung dari garis tersebut telah diketahui.

Contoh tampilan program:
```
Koordinat-x titik pangkal = 2
Koordinat-y titik pangkal = 3

Koordinat-x titik ujung = 6
Koordinat-y titik ujung = 6

Kemiringan garis = 0.75
Panjang garis = 5.0
```

````python
# program ini disusun untuk menentukan kemiringan garis dan panjang garis, di mana koordinat masing-masing titik diketahui
import math
x_1 = 2        # Koordinat-x titik pangkal
y_1 = 3        # Koordinat-y titik pangkal
x_2 = 6        # Koordinat-x titik ujung
y_2 = 6        # Koordinat-y titik ujung

# perhitungan
m = (y_2 - y_1) / (x_2 - x_1)                   # Mencari Kemiringan garis
l = math.sqrt((x_2 - x_1)**2 + (y_2 - y_1)**2)  # Mencari Panjang garis

# tampilkan hasil
print('Koordinat-x titik pangkal =', x_1)
print('Koordinat-y titik pangkal =', y_1)
print('Koordinat-x titik ujung =', x_2)
print('Koordinat-y titik ujung =', y_2)
print('Kemiringan garis', m)
print('Panjang garis =', l)
````
