### <span style="color:blue">Kerja #1: Tahun Kabisat &#8594; 15 poin</span>

Suatu tahun disebut tahun <b>kabisat</b> jikalau terdapat 366 hari di dalam tahun tersebut. Tahun yang lazim dan tidak kabisat mengandung hanya 365 hari. Untuk menentukan apakah suatu tahun termasuk tahun kabisat atau tidak, dapat dipergunakan rumus sebagai berikut:

> Bilangan tahun kabisat seharusnya habis dibagi 4 namun tidak termasuk tahun abad (kelipatan 100) yang tidak habis ketika dibagi 400.

Sebagai contoh:

- tahun 1600 dan 2000 adalah tahun abad yang termasuk tahun kabisat, sedangkan
- tahun 1800 dan 1900 adalah tahun abad yang tidaklah termasuk tahun kabisat.

<span style="color:red">&#9881;</span> Pada sel di bawah ini, tulislah program Python <b>lengkap</b> bernama `TahunKabisat.py` yang menentukan apakah tahun yang diberikan pengguna program melalui papan-ketik (<i>keyboard</i>) tergolong tahun kabisat atau tidak.

```python
# TahunKabisat.py: Program Python lengkap yang menentukan apakah tahun yang diberikan pengguna
#                  program melalui papan-ketik (keyboard) tergolong tahun kabisat atau tidak

# Masukkan perintah untuk menginput tahun
Tahun = input('Tahun =  ')
Tahun = float(Tahun)

# Masukkan perintah percabangan
if (Tahun % 4) == 0:
    if (Tahun % 100) == 0:
        if (Tahun % 400) == 0:
            print('Tahun tersebut merupakan tahun kabisat')
        else:
            print('Tahun tersebut bukan merupakan tahun kabisat')
    else:
        print('Tahun tersebut bukan merupakan tahun kabisat')
else:
    print('Tahun tersebut bukan merupakan tahun kabisat')
```

### <span style="color:blue">Kerja #2: Index Massa Tubuh &#8594; 25 poin</span>

Index massa tubuh atau <i>body mass index</i> (BMI) adalah suatu nilai yang menentukan apakah perbandingan antara berat badan dan tinggi badan seseorang dapat dianggap sehat atau tidak. BMI dihitung dengan cara:

> Berat badan (dalam satuan pound) dikalikan tetapan 750 dan hasilnya dibagi dengan kuadrat tinggi badan (dalam satuan inci).

Jika nilai BMI seseorang terletak di antara rentang dari 19 hingga 25 (termasuk nilai 19 dan nilai 25) maka orang tersebut dapat dikatakan termasuk sehat.

<span style="color:red">&#9881;</span> Pada sel di bawah ini, tulislah program Python <b>lengkap</b> bernama `bmi.py` yang menentukan apakah index massa tubuh dari pengguna program termasuk sehat atau tidak. Program ini

- meminta pengguna untuk memberikan berat badannya dalam satuan kilogram;
- meminta pengguna untuk memberikan tinggi badannya dalam satuan sentimeter;
- menghitung nilai BMI dari pengguna berdasarkan informasi yang diberikannya; serta
- menampilkan nilai BMI dan suatu pesan yang menginformasikan apakah nilai BMI tersebut berada di bawah, di dalam atau di atas rentang nilai BMI yang sehat.

<u>Petunjuk</u>: Gunakan konversi satuan pengukuran berikut: 1 kg = 2.20 lb dan 1 cm = 0.39 in.

```python
# bmi.py: Program Python lengkap yang menentukan apakah index massa tubuh dari pengguna
#         program termasuk sehat atau tidak

# Masukkan perintah untuk input berat badan
b = input('Masukkan berat badan Anda (kg) = ')
b = float(b)

# Konversi satuan dari kg ke pound
b2 = b * 2.20

# Masukkan perintah untuk input tinggi badan
t = input('Masukkan tinggi badan Anda (cm) = ')
t = float(t)

# Konversi satuan dari cm ke in
t2 = t * 0.39

# Masukkan rumus BMI
BMI = (b2 * 750) / (t2**2)

# Perintah percabangan
if BMI >= 19 and BMI <= 25:
    print('Nilai BMI adalah %g, selamat tubuh Anda dapat dinyatakan sehat !' % BMI)
else:
    print('Nilai BMI adalah %g, sayang sekali tubuh Anda belum bisa dikatakan sehat' % BMI)
```

Pada akhir kerjanya, fungsi `faktorBilangan()` mengembalikan suatu <i>list</i> yang berisikan semua faktor dari parameternya.

```python
# faktorBilangan : Definisi fungsi Python yang menentukan semua faktor dari suatu
#                  bilangan ulat

# Buat definisi fungsi
def faktorBilangan(x):
    print('Faktor dari', x, 'adalah')
    for y in range(1, x + 1): # Perintah perulangan
        if x % y == 0:
            print(y)
```

### <span style="color:blue">Kerja #4: Hitung Polinomial &#8594; 35 poin</span>

Polinomial dapat diartikan sebagai ekspresi matematika yang melibatkan suatu jumlahan suku-suku berpangkat dari satu variabel atau lebih yang masing-masing dikalikan dengan koefisien tertentu. Bentuk umum dari polinomial dengan satu variabel $x$ adalah

$$ a_n \, x^n + a_{n - 1} \, x^{n - 1} + \cdots + a_2 \, x^2 + a_1 \, x + a_0 \quad\quad\quad (1) $$

Polinomial pada Persamaan $(1)$ juga dapat dinyatakan sebagai $p(x)$ yang disusun dari $n + 1$ akar atau <i>roots</i>: $r_n, r_{n - 1}, \cdots, r_2, r_1, r_0$, sebagai berikut:

$$ p(x) = \prod_{i = 0}^n (x - r_i) = (x - r_0)(x - r_1)(x - r_2) \cdots (x - r_{n - 1})(x - r_n) \quad\quad\quad (2) $$


<span style="color:red">&#9881;</span> Pada sel di bawah ini, tulislah definisi fungsi Python bernama `polinom(x, akar)` yang menerima argumen-argumen: `x` dan suatu <i>list</i> `akar` untuk menghasilkan nilai kembalian dari polinomial $p(x)$ seperti pada Persamaan $(2)$.

```python
# polinom(x, akar) : Definisi fungsi Python yang menerima argumen-argumen x dan suatu list
#                    akar untuk menghasilkan kembalian dari polinomial p(x)
#                    seperti pada persamaan dua

# Definisi fungsi
def polinom(x, akar):
    p_x = 1
    for i in akar:
        p_x = p_x * (x-1)
    return p_x
```

- satu-per-satu sebanyak `n` bilangan bulat atau pecahan yang merupakan $n$ akar $r_i$ dari polinomial pada Persamaan $(2)$; dan
- satu bilangan bulat atau pecahan yang merupakan nilai variabel $x$ pada Persamaan $(1)$ dan $(2)$.
