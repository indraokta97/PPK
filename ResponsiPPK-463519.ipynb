<span style="color:red">&#9881;</span> &nbsp; Pada sel di bawah ini, tulislah program Python <b>lengkap</b> bernama `GrafikDataBerkas.py` yang melakukan 2 langkah berikut:

- membaca data kerapatan udara dari dalam berkas `density_air.dat`; dan

Jangan lupa untuk melengkapi dekorasi yang sesuai terhadap grafik yang anda buat.

```python
# GrafikDataBerkas.py: Program Python lengkap yang membaca data kerapatan udara dari dalam berkas density_air.dat 
#                      dan melukiskan grafik rapat udara vs suhu lingkungan dengan lambang lingkaran kecil terpisah untuk
#                      tiap titik data.

# Import pustaka yang diperlukan
import numpy as np
import matplotlib.pyplot as plt

# Membuka dan membaca berkas "density_air.dat"
berkas = open('density_air.dat','r') 
baca = berkas.readlines()

# Memuat list kosong 
list_x = [] # Suhu
list_y = [] # Kerapatan

# Isi list kosong tersebut dengan menggunakan perintah kalang for
for i in range(3,12):
    data = baca[i].split()
    list_x.append(float(data[0]))
    list_y.append(float(data[1]))

# Menutup berkas
berkas.close()

# Membuat larik berdasarkan list suhu yang diketahui
larik_x = np.array(list_x)  
larik_y = np.array(list_y)  

# Insert perintah gambar
gambar = plt.figure()

# Tambahkan dekorasi pada curva
plt.title('Grafik Kerapatan Udara vs Suhu Lingkungan') # Judul grafik
plt.xlabel('Suhu lingkungan (℃)')                     # Keterangan x
plt.ylabel('Kerapatan Udara (kg/m³)')                  # Keterangan y
plt.plot(larik_x, larik_y, 'bo')                       # Bentuk bulat
plt.legend(['Kerapatan Udara vs Suhu lingkungan'])     # Legenda curva

# Tampilkan curva
plt.show() 
```

Pustaka NumPy memiliki fungsi `polyfit(x, y, deg)` untuk menentukan pencocokan (<i>fitting</i>) polinomial terbaik dengan orde `deg` terhadap sekumpulan titik data yang diberikan oleh argumen larik `x` dan larik `y`. Fungsi `polyfit()` ini mengembalikan suatu <i>list</i> yang berisikan koefisien-koefisien dari polinomial hasil pencocokan, dengan

- elemen pertamanya adalah koefisien dari suku polinomial dengan pangkat tertingi; dan
- elemen terakhirnya adalah koefisien dari suku polinomial yang berupa tetapan.

Sebagai contoh: Jika diberikan sekumpulan titik `(x, y)` maka perintah `polyfit(x, y, 1)` mengembalikan <i>list</i> koefisien `[a, b]` yang berperan dalam polinomial berorde-1 atau persamaan garis lurus:

$$ y_{fit} = a \cdot x + b \quad\quad (1) $$

dengan $y_{fit}$ adalah hasil pencocokan terbaik terhadap titik `y` pada nilai `x` yang sama.

<span style="color:red">&#9881;</span> &nbsp; Pada sel di bawah ini, tulislah program Python <b>lengkap</b> bernama `CocokDataBerkas.py` yang melakukan algoritma sebagai berikut:

4. Lukiskan satu grafik yang berisikan tiga kurva:
    - rapat udara vs suhu lingkungan dengan lambang lingkaran kecil yang terpisah untuk setiap data;
    - rapat udara hasil pencocokan polinomial berorde-1 vs suhu lingkungan dengan lambang garis berwarna biru; dan
    - rapat udara hasil pencocokan polinomial berorde-2 vs suhu lingkungan dengan lambang garis berwarna merah.

Jangan lupa untuk melengkapi dekorasi yang sesuai terhadap grafik yang anda buat.

Suatu piranti GPS (<i>Global Positioning System</i>) mengukur posisi pengguna `(x, y)` pada setiap `s` detik. Data posisi pada suatu perjalanan dapat disimpankan ke dalam suatu berkas, seperti berkas `pos.dat`. Berkas ini berisikan:

- nilai `s` (detik) pada baris pertama; dan
- koordinat $x$ dan $y$ dari posisi yang terukur GPS pada baris-baris selanjutnya.

Jika $x(t)$ dan $y(t)$ adalah koordinat posisi sebagai fungsi waktu maka kecepatan pada arah $x$ dan $y$ masing-masing adalah

$$ v_x = \frac{dx}{dt} \quad \text{dan} \quad v_y = \frac{dy}{dt} \quad\quad (2) $$

Karena nilai-nilai $x$ dan $y$ hanya terukur pada $n$ waktu yang diskrit (terpisah-pisah):

$$ t_k = k \cdot s,\quad \text{dengan} \quad k = 0, 1, 2, \dots, n - 1 \quad\quad (3) $$

maka rumus kecepatan $v_x$ dan $v_y$ dapat didekati menggunakan pendiferensialan numerik sebagai berikut:

$$ v_x(t_k) \approx \frac{x(t_{k+1}) - x(t_k)}{s} \quad \text{dan} \quad v_y(t_k) \approx \frac{y(t_{k+1}) - y(t_k)}{s} \quad\quad (4) $$
