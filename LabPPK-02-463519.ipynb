### <span style="color:blue">Kerja #1: Konversi Satuan Panjang &#8594; 25 poin</span>

Pada saat ini, ada beberapa sistem satuan (<i>unit system</i>) yang sudah dikembangkan dan dipergunakan secara meluas, seperti <i>International Systems of Units</i> (SI), sistem MKS (meter-kilogram-sekon), sistem CGS (sentimeter-gram-sekon), sistem Imperial (British) dan sebagainya. Kerapkali, agar berhasil dalam melakukan penyelesaian masalah keteknikan diperlukan pengubahan atau konversi satuan dari suatu sistem satuan menjadi sistem lainnya.

Sebagai contoh: satuan-satuan panjang yang didefinisikan pada sistem Imperial adalah

- 1 inci = 2.54 cm
- 1 kaki = 12 inci
- 1 yar = 3 kaki
- 1 mil = 1760 yar

<span style="color:red">&#9881;</span> Pada sel di bawah ini, susunlah program Python __lengkap__ bernama `KonversiPanjang.py` yang melakukan konversi satuan panjang dari meter menjadi satuan-satuan: inci, kaki, yar dan mil. Ujilah kebenaran progam anda dengan menggunakan suatu panjang dalam meter berikut ini: 
<p>
<center> 640 m = 25196.85 inci = 2099.74 kaki = 699.91 yar = 0.3977 mil</center>

````python
# KonversiPanjang.py: Program yang melakukan konversi satuan panjang dari meter menjadi satuan
#                     inci, kaki, yar, dan mil

# Diketahui
# 1 inci = 2.54 cm
# 1 kaki = 12 inci
# 1 yar = 3 kaki
# 1 mil = 1760 yar

# Konversi satuan panjang
meter = 640                   # Panjang dalam satuan meter
inci = meter * 100 / 2.54     # Konversi satuan meter menjadi satuan inci (dikali 100 karena ubah m menjadi cm terlebih dulu)
kaki = inci / 12              # Konversi satuan inci menjadi satuan kaki
yar = kaki / 3                # Konversi satuan kaki menjadi satuan yar
mil = yar / 1760              # Konversi satuan yar menjadi satuan mil

# Tampilkan program konversi panjang dari meter menjadi beberapa satuan
print('%g meter = %.2f inci = %.2f kaki = %.2f yar = %.4f mil' % (meter, inci, kaki, yar, mil))
````

### <span style="color:blue">Kerja #2: Penanaman Modal &#8594; 25 poin</span>

Salah satu dari bebagai jasa yang disediakan oleh perbankan adalah penanaman modal. Dengan memanfaatkan jasa ini, seorang penanam modal atau investor dapat meningkatkan nilai modal-awal yang dimilikinya menjadi berlipat kali setelah melewati masa penanaman selama jangka waktu tertentu.

Rumus penanaman modal yang sering dipergunakan adalah

$$ F = A \, \left( 1 + \frac{p}{100} \right)^n $$

dengan

- $F$ = nilai modal pada masa depan
- $A$ = nilai modal-awal (<i>pincipal</i>)
- $p$ = suku-bunga bank dalam persen per tahun
- $n$ = masa penanaman dalam satuan tahun

<span style="color:red">&#9881;</span> Pada sel di bawah ini, susunlah program Python __lengkap__ bernama `PenanamanModal.py` yang menghitung berapa banyak modal sebesar Rp10.000.000,- meningkat nilainya setelah ditanam selama 3 tahun pada suatu bank yang memiliki suku-bunga sebesar 5 persen per tahun.

````python
# PenanamanModal.py : Program yang menghitung berapa banyak modal sebesar Rp10.000.000,- 
#                     meningkat nilainya setelah ditanam selama 3 tahun pada suatu bank yang memiliki 
#                     suku-bunga sebesar 5 persen per tahun.

# Diketahui nilai dari
A = 10000000       # Nilai modal awal 
p = 5              # Suku bunga bank dalam persen pertahun
n = 3              # masa penanaman dalam satuan tahun

# Masukkan ke dalam rumus
F = A * (1 + p / 100)**n   # Nilai modal pada masa depan

# Tampilkan program
print('Nilai modal pada masa depan (F) =', F)
````

### <span style="color:blue">Kerja #4: Persamaan Kuadrat &#8594; 25 poin</span>

Suatu persamaan kuadrat dengan bentuk umum:

$$ a \, x^2 + b \, x + c = 0 $$

memiliki dua akar atau penyelesaian. Kedua akar tersebut dapat dihitung dengan menggunakan rumus:

$$ x_1 = \frac{-b + \sqrt{b^2 - 4 \, a \, c}}{2 \, a}, \quad x_2 = \frac{-b - \sqrt{b^2 - 4 \, a \, c}}{2 \, a} $$

Seorang pemrogram hendak mengimplementasikan penyelesaian persamaan kuadrat dengan kode-sumber sebagai berikut:
```
a = 2;  b = 1;  c = 2
from math import sqrt
q = b * b - 4 * a * c
q_sr = sqrt(q)
x1 = (-b + q_sr) / 2 * a
x2 = (-b - q_sr) / 2 * a
print(x1, x2)
```

<span style="color:red">&#9881;</span> Pada sel di bawah ini, tulislah semua sumber kesalahan pada kode-sumber di atas dan kemudian perbaikilah sehingga program tersebut dapat berjalan dengan benar. Ujilah kebenaran program anda dengan menyulihkan nilai-nilai akar yang diperoleh ke dalam persamaan kuadrat terkait.

````python
# Persamaan kuadrat

# Sumber kesalahan : Dari program di atas, kesalahan terdapat pada kode sumber "from math import sqrt".
#                    Modul "math" tidak dapat berjalan karena nilai b^2 < 4*a*c sehingga menyebabkan
#                    b^2 - 4*a*c bernilai negatif yang mana akan terjadi error apabila 
#                    dimasukkan ke fungsi akar. 

# Perbaikan : Perbaikan kode pemrograman tersebut dapat dilakukan dengan menggunakan modul matematika
#             yang kompleks, yaitu modul "cmath". 

# Jalankan program yang sudah diperbaiki
a = 2;  b = 1;  c = 2
from cmath import sqrt
q = b * b - 4 * a * c
q_sr = sqrt(q)
x1 = (-b + q_sr) / (2 * a)
x2 = (-b - q_sr) / (2 * a)
print(x1, x2)
````
