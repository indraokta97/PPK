### <span style="color:blue">Kerja #1: Sisi Miring Segitiga &#8594; 15 poin</span>

Setiap segitiga siku-siku memiliki 3 sisi yang dinamakan

- alas dengan panjang $x$,
- tinggi dengan panjang $y$, dan
- sisi miring (<i>hypotenuse</i>) dengan panjang $r$.

Hubungan di antara panjang ketiga sisi tersebut dinyatakan sebagai Teorema Pythagoras:
<p>
$$ r = \sqrt{x^2 + y^2} $$

<span style="color:red">&#9881;</span> Pada sel di bawah ini, tulislah definisi fungsi `sisi_miring()` yang menerima dua argumen: panjang sisi-sisi alas dan tinggi dari suatu segitiga siku-siku dan mengembalikan panjang sisi-miring dari segitiga tersebut. Periksalah kebenaran kerja fungsi yang anda tulis dengan memanggil namanya dan memberikan dua nilai argumen yang diperlukannya. Bandingkan hasil perhitungan fungsi buatan anda tersebut dengan perhitungan rumus yang sama menggunakan kalkulator ilmiah.

```python
# sisi_miring() : Program yang merupakan sebuah fungsi yang menerima dua argumen
#                 yaitu panjang sisi-sisi alas dan tinggi dari suatu segitiga siku-siku
#                 dan mengembalikan panjang sisi miring dari segitiga tersebut

# Panggil pustaka
from math import sqrt

# Tuliskan fungsi
def sisi_miring(x,y):
    r = sqrt(x**2 + y**2)
    return r

# Misalnya diketahui
x = 5                  # Panjang alas segitiga siku-siku
y = 12                 # Tinggi segitiga siku-siku
r = sisi_miring(x,y)   # Sisi miring segitiga siku-siku

# Tampilkan program
print('x = %g y = %g r = %g' % (x, y, r))

# Bandingkan dengan perhitungan kalkulator
print('Perhitungan dengan kalkulator ilmiah menunjukkan r = 13')
print('Maka, hasil perhitungan dengan program dan kalkulator ilmiah adalah sama')
```

Pada kebanyakan program komputer, fungsi-fungsi matematika seperti di bawah ini:

$$ g(t) = e^{-t} \, sin(\pi \, t) \quad\quad (1)$$
$$ h(t) = e^{-a \, t} \, sin(\pi \, t) \quad\quad (2)$$

dapat di-implementasi-kan sebagai suatu definisi fungsi dalam bahasa pemrograman tertentu. Definisi fungsi tersebut mungkin tidak disertai dengan suatu argumen dan mungkin pula disertai dengan satu atau beberapa argumen yang diperlukan agar fungsi tersebut dapat bekerja secara benar.

<span style="color:red">&#9881;</span> Pada sel di bawah ini, tulislah definisi fungsi Python `g()` yang mewujudkan fungsi $g(t)$ di atas [Persamaan 1]. Laju uji kebenaran fungsi buatan anda dengan memanggil fungsi `g()` dan secara berturut-turut memberikan nilai-nilai argumen $t = 0, 0.25, 0.5, 0.75$ dan $1$.

<b>Petunjuk</b>: Periksa hasil pengujian fungsi buatan anda dengan tampilan berikut:
![hasil-fungsi-g()](hasil-fungsi-g.png)

```python
# Fungsi Python g() : Program yang mewujudkan fungsi g(t) pada persamaan satu.
#                     dengan pemanggilan fungsi g() dengan nilai t = 0, 0.25, 0.5, 0.75, dan 1

# Panggil pustaka
from math import sin, pi, e

# Tuliskan fungsi
def g(t) :
    g = e**(-t) * sin(pi * t)
    return g

# Nilai t
t1 = 0 ; t2 = 0.25; t3 = 0.5 ; t4 = 0.75 ; t5 = 1

# Tampilkan hasil
print('g(0)    = %g' %g(t1))
print('g(0.25) = %g' %g(t2))
print('g(0.5)  = %g' %g(t3))
print('g(0.75) = %g' %g(t4))
print('g(1)    = %g' %g(t5))
```

Laju uji kebenaran fungsi buatan anda dengan memanggil fungsi `h()` dan secara berturut-turut memberikan nilai-nilai argumen $t = 0, 0.25, 0.5, 0.75$ dan $1$.

```python
# Fungsi Python h() : Program yang mewujudkan fungsi h(t) di atas (persamaan 2)
#                     dengan parameter a = 10 ditetapkan sebagai suatu fungsi global.
#                     Pemanggilan fungsi h() dan nilai t = 0, 0.25, 0.5, 0.75, dan 1.

# Panggil pustaka
from math import sin, pi, e

# Tuliskan fungsi
def h(t) :
    a = 10
    h = e**(-a * t) * sin(pi * t) 
    return h

# Nilai t
t1 = 0 ; t2 = 0.25; t3 = 0.5 ; t4 = 0.75 ; t5 = 1

# Tampilkan hasil
print('h(0)    = %g' %h(t1))
print('h(0.25) = %g' %h(t2))
print('h(0.5)  = %g' %h(t3))
print('h(0.75) = %g' %h(t4))
print('h(1)    = %g' %h(t5))
```

Bayangkan anda adalah sang pakar merebus telur yang paling sakti sedunia! Rumah makan anda selalu penuh dengan pengunjung yang ingin menyantap telur rebus buatan anda. Ketika pesanan telur rebus bertambah, anda memerlukan waktu yang lebih lama untuk merebus karena anda hanya memiliki satu periuk saja untuk melaksanakan pekerjaan anda. Berapa lamakah waktu merebus telur yang anda butuhkan?

<span style="color:red">&#9881;</span> Pada sel di bawah ini, tulislah definisi fungsi Python bernama `waktuRebus()` yang

- menerima bilangan $n > 0$ yang merupakan cacah telur yang harus direbus, dan
- mengembalikan waktu yang diperlukan (dalam menit) agar semua telur dapat terebus hingga matang.

Beberapa aturan yang perlu diperhatikan dalam membuat definisi fungsi ini adalah

- cacah telur yang dapat dimasukkan ke dalam periuk sebanyak-banyaknya 8 telur saja,
- diperlukan waktu selama 5 menit untuk merebus satu telur hingga matang,
- anggaplah air di dalam periuk selalu mendidih sepanjang waktu sehingga tidak diperlukan masa pemanasan awal, dan
- anggaplah tidak diperlukan waktu untuk memasukkan dan mengeluarkan telur ke/dari dalam periuk.

```python
# waktuRebus() : Fungsi yang menunjukkan waktu rebus telur

t = 5 
def waktuRebus(n):
    waktu = t + ((n - 1) // 8) * t
    return waktu
```

### <span style="color:blue">Kerja #4: Konversi Suhu &#8594; 15 poin</span>

Rumus konversi suhu dari satuan Fahrenheit menjadi satuan Celsius dapat dituliskan sebagai suatu fungsi matematika sebagai berikut:

$$ C(F) = \frac{5}{9} \, (F - 32) $$


<span style="color:red">&#9881;</span> Pada sel di bawah ini, tulislah definisi fungsi Python bernama `C(F)` yang mengimplementasikan fungsi matematika di atas. Dan tulis juga definisi fungsi kebalikan yang bernama `F(C)` yang melakukan konversi suhu dari satuan Celsius menjadi satuan Fahrenheit.

```python
# Fungsi Python bernama C(F) adalah konversi suhu dari Fahrenheit ke Celsius
def C(F):
    C = 5 / 9 * (F - 32)
    return C

# Fungsi Python bernama F(C) adalah konversi suhu dari Celsius ke Fahrenheit
def F(C):
    F = 9 / 5 * C + 32
    return F
```

### <span style="color:blue">Kerja #5: Jumlah Deret &#8594; 20 poin</span>

Pada proses perhitungan optimalisasi, kerapkali diperlukan penjumlahan dari sederet nilai dan penjumlahan kuadrat dari sederet nilai. Kedua perhitungan tersebut dapat dinyatakan sebagai fungsi matematika sebagai berikut:

$$ sum_1(n) = \sum_{i = 1}^n i \quad\quad (3)$$
$$ sum_2(n) = \sum_{i = 1}^n i^2 \quad\quad (4)$$

<span style="color:red">&#9881;</span> Pada sel di bawah ini, tulislah definisi fungsi Python `sum1()` yang mewujudkan fungsi $sum_1(n)$ di atas [Persamaan (3)]. Gunakan kalang `while` dalam melakukan penjumlahan yang diperlukan. Ujilah kebenaran definisi fungsi buatan anda dengan memanggilnya dan membandingkan hasil penjumlahan dengan yang dihasilkan memakai kalkulator ilmiah.

```python
# sum1() : Fungsi yang mewujudkan fungsi sum1(n) (persamaan 3)

# Tuliskan fungsi dan gunakan kalang while
def sum1(n):
    kosong = []
    i = 0
    while i < n + 1:
        kosong.append(i)
        i = i + 1
    return sum(kosong)

# Misalkan
n = 100

# Maka
print('Hasil perhitungan dengan menggunakan program apabila n = 100 adalah ', sum1(n))
print('Hasil perhitungan dengan menggunakan kalkulator ilmiah apabila n = 100 adalah 5050')
print('Jadi, hasil perhitungan menggunakan program dan kalkulator adalah sama')
```

<span style="color:red">&#9881;</span> Pada sel di bawah ini, tulislah definisi fungsi Python `sum2()` yang mewujudkan fungsi $sum_2(n)$ di atas [Persamaan (4)]. Gunakan kalang `for` dalam melakukan penjumlahan yang diperlukan. Ujilah kebenaran definisi fungsi buatan anda dengan memanggilnya dan membandingkan hasil penjumlahan dengan yang dihasilkan memakai kalkulator ilmiah.

```python
# sum1() : Fungsi yang mewujudkan fungsi sum2(n) (persamaan 4)

# Tuliskan fungsi dan gunakan kalang for
def sum2(n):
    kosong = []
    i = 0
    for i in range(n + 1):
        i = i**2
        kosong.append(i)
    return sum(kosong)

# Misalkan
n = 100

# Maka
print('Hasil perhitungan dengan menggunakan program apabila n = 100 adalah ', sum2(n))
print('Hasil perhitungan dengan menggunakan kalkulator ilmiah apabila n = 100 adalah 338350')
print('Jadi, hasil perhitungan menggunakan program dan kalkulator adalah sama')
```
