### <span style="color:blue">Kerja #1: Jumlah Bilangan Bulat &#8594; 20 poin</span>

Bilangan bulat adalah bilangan yang tidak melibatkan bagian pecahan. Sederetan bilangan bulat dapat dibangkitkan dengan membilang secara berturutan:

$$ 0, 1, 2, 3, 4, 5, 6, \dots $$

Jumlah sederetan $n$ bilangan bulat yang pertama sudah lama diteliti dan sudah ada rumus terkenal dengan jumlah tersebut, yakni

$$ 1 + 2 + 3 + 4 + 5 + 6 + \dots + n = \frac{n (n + 1)}{2} $$

<span style="color:red">&#9881;</span> Pada sel di bawah ini, susunlah program Python __lengkap__ bernama `JumlahBilBulat.py` yang menggunakan konstruksi kalang untuk menghitung jumlah bilangan bulat berturutan dimulai dari 1 hingga (termasuk) $n$. Tetapkan nilai $n$ pada awal program anda. Dan untuk membuktikan kebenarannya, bandingkan hasil penjumlahan tersebut dengan jumlah yang diperoleh dari rumus: $n (n+1) / 2$.

```python
# JumlahBilBulat.py: Program yang menggunakan konstruksi kalang untuk menghitung jumlah 
#                    bilangan bulat berturutan dimulai dari 1 hingga (termasuk) n.

# Tetapkan nilai suku terakhir
n = 20                 # suku terakhir

# Hitung dengan menggunakan kalang while
x = 1                  # suku pertama
dx = 1                 # pertambahan nilai
hasil = 0
while x <= 20:
    hasil += x
    x += dx
print('Jumlah deret dari 1 hingga 20 menggunakan program =', hasil)

# Hitung dengan menggunakan rumus
jumlah_deret = (n * (n + 1)) / 2
print('Jumlah deret dari 1 hingga 20 menggunakan rumus =', jumlah_deret)
print('maka,')
print('Jumlah deret dengan program = Jumlah deret dengan rumus, yaitu', 210)
```

### <span style="color:blue">Kerja #2: <i>List</i> Bilangan Prima &#8594; 15 poin</span>

Bilangan prima adalah bilangan bulat lebih besar daripada 1 dan hanya habis dibagi (tanpa sisa) oleh dirinya sendiri dan satu.

<span style="color:red">&#9881;</span> Pada sel di bawah ini, susunlah program Python __lengkap__ bernama `BilanganPrima.py` yang memberikan kepada variabel `prima` berupa suatu <i>list</i> yang berisikan nilai-nilai bilangan: $2, 3, 5, 7, 11$ dan $13$.

- Gunakanlah kalang `for` untuk menampilkan satu-per-satu elemen <i>list</i> tersebut.
- Berikan nilai $17$ kepada variabel `p`, lalu imbuhkan variabel ini ke penghujung <i>list</i> `prima`.
- Gunakanlah kalang `while` untuk menampilkan satu-per-satu elemen <i>list</i> yang baru diubah tesebut.

```python
# BilanganPrima.py: Program yang menamppilkan bilangan prima berisikan 2, 3, 5, 7, 11, dan 13

# Bilangan prima sebelum ditambahkan nilai baru
BilanganPrima = [2, 3, 5, 7, 11, 13]
print('Elemen bilangan prima sebelum ditambahkan nilai baru:')
for p in BilanganPrima :
    print ('Elemen list =', p)
print()

# Tambahkan nilai baru, misal x
x = 17
BilanganPrima.append(x)

# Tampilkan bilangan prima setelah ditambahkan nilai x
print('Elemen bilangan prima setelah ditambahkan nilai baru:')
i = 0
while i !=  7:
    print('Elemen list =', BilanganPrima [i])
    i = i+1
```

### <span style="color:blue">Kerja #3: Deret Bilangan Ganjil &#8594; 15 poin</span>

Bilangan ganjil (<i>odd</i>) adalah bilangan bulat yang tidak habis dibagi dua.

<span style="color:red">&#9881;</span> Pada sel di bawah ini, susunlah program Python __lengkap__ bernama `BilanganGanjil.py` yang membangkitkan suatu deret yang berisikan semua bilangan ganjil dimulai dari 1 hingga (termasuk) `n`. Tetapkan nilai `n` pada awal program anda dan gunakan kalang `while` untuk menentukan masing-masing elemen deret tersebut.

<b>Petunjuk</b>: Jika `n` adalah suatu bilangan genap maka pastikan elemen terbesar di dalam deret tersebut adalah `n - 1`. 

```python
# Bilanganganjil.py : Program yang membangkitkan suatu deret yang berisikan semua
#                     bilangan ganjil dimulai dari 1 hingga (termasuk) n.

# Petunjuk = n merupakan suatu bilangan genap
n = 20             # suku terakhir (batas atas)

# Suku pertama dan selisih
x = 1              # suku pertama (batas bawah)
dx = 2             # selisih antar suku

# List yang tidak diisi (untuk membantu perhitungan menggunakan kalang while)
BilanganGanjil = []

#perhitungan dengan kalah while
while x <= n:
    BilanganGanjil.append(x)
    x += dx
    
print('Deret bilangan ganjil', BilanganGanjil)
print('Banyak bilangan ganjil dalam deret tersebut adalah', len(BilanganGanjil), 'nilai')
print('Elemen terbesar dalam deret bilangan ganjil adalah', (n - 1))
```

### <span style="color:blue">Kerja #4: Tabel Konversi Suhu &#8594; 25 poin</span>

Banyak orang yang menggunakan rumus pendekatan atau aproksimasi untuk secara cepat melakukan konversi satuan suhu Fahrenheit menjadi Celsius dengan rumus:

$$ C \approx \hat{C} = \frac{F - 30}{2} $$

<span style="color:red">&#9881;</span> Pada sel di bawah ini, susunlah program Python __lengkap__ bernama `TabelAproksimasi.py` yang menampilkan suatu tabel konversi satuan suhu yang rapi. Tabel tersebut terdiri dari tiga kolom, yakni $F$, $C$ dan $\hat{C}$. Kolom pertama tabel tersebut berisikan nilai-nilai suhu $F$ yang teratur: $0, 10, 20, \dots, 100$.

```python
# TabelAproksimasi.py: Program yang menampilkan suatu tabel konversi satuan suhu yang rapi. 
#                      Tabel tersebut terdiri dari tiga kolom, yakni  𝐹 ,  𝐶  dan  𝐶̂. 
#                      Kolom pertama tabel tersebut berisikan nilai-nilai suhu  𝐹  
#                      yang teratur:  0,10,20,…,100 .

# Tampilkan awal tabel dan besaran yang diperlukan
print('------------------------------')      # kepala tabel
F = 0                                        # nilai awal suhu dalam Fahrenheit
dF = 10                                      # pertambahan suhu Fahrenheit

# List yang tidak diisi (untuk membantu perhitungan menggunakan kalang while)
Fahrenheit = []

# Ulangi perhitungan dengan kalang while
print('F          C                 C^')            # keterangan kolom
while F <= 100:                                     # kepala kalang dengan kondisi tertentu
    Fahrenheit.append(F)                            # penambahan nilai F
    F += dF                                         # nilai F dalam kalang
    
# Konversi suhu dari F ke C dan C^
for F in Fahrenheit:
    C = 5 / 9 * (F - 32)                     # nilai C 
    C1 = (F - 30) / 2                        # nilai C^                            
    print(F, '|', C, '|', C1) 
print('------------------------------')      # ekor tabel
```

### <span style="color:blue">Kerja #5: Tenaga Atom Hidrogen &#8594; 25 poin</span>

Aras (<i>level</i>) tenaga ke-$n$ dari elektron dalam atom Hidrogen ditentukan dengan rumus berikut:

$$ E_n = -\frac{m_e \, e^4}{8 \, \epsilon_0^2 \, h^2} \cdot \frac{1}{n^2} $$

dengan

- $m_e = 9.1094 \times 10^{-31}$ kg : massa elektron
- $e = 1.6022 \times 10^{-19}$ C : muatan elementer
- $\epsilon_0 = 8.8542 \times 10^{-12}$ C$^2$s$^2$kg$^{-1}$ : pemitivitas listrik dari ruang vakum
- $h = 6.6261 \times 10^{-34}$ Js : tetapan Planck

<span style="color:red">&#9881;</span> Pada sel di bawah ini, susunlah program Python __lengkap__ bernama `ArasTenagaHidrogen.py` yang menampilkan suatu tabel aras-aras tenaga elektron dalam atom Hidrogen untuk $n = 1, 2, 3, \dots, 20$. 

```python
# ArasTenagaHidrogen.py: Program yang menampilkan suatu tabel aras-aras 
#                        tenaga elektron dalam atom Hidrogen untuk n = 1, 2, 3, ..., 20.

# Diketahui
m_e = 9.1094 * 10**-31        # massa elektron (kg)
e = 1.6022 * 10**-19          # muatan elementer (C)
e_o = 8.8542 * 10**-12        # pemitivitas listrik ruang vakum (C^2s^2kg^-1)
h = 6.6261 * 10**-34          # tetapan Plank (Js)

# Rumus aras dan tampilkan tabel n dan E
print('Tabel aras-aras tenaga elektron dalam atom Hidrogen')
print(' n |     E')
for n in range(1, 21, 1):
    E = - (m_e * e**4) / (8 * e_o**2 * h**2) * 1 / n**2
    print('%2d'  %(n), '|' , '%6.4g'  %(E))
```
