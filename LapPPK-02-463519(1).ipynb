### <span style="color:blue">Tugas #1: <i>Debugging</i> &#8594; 30 poin</span>

Penentuan sumber kesalahan dan perbaikannya (<i>debugging</i>) merupakan suatu keahlian yang penting dalam pemrograman komputer. Ada dua golongan kesalahan pada suatu program, yakni

- <b>Kesalahan sintaksis</b> atau kesalahan penulisan yang membuat program tidak berjalan hingga tuntas dan <i>interpreter</i> atau <i>compiler</i> menampilkan pesan kesalahan serta kemungkinan letak sumber kesalahan. Pemrogram harus mampu menafsirkan arti dari pesan tersebut, memastikan letak sumber kesalahan dan memperbaiki kesalahan yang dimaksud.
- <b>Kesalahan logika</b> yang tidak menimbulkan pesan kesalahan namun program menghasilkan output yang salah atau bukan seperti yang diinginkan. Berbekalkan algoritma yang mendasari program, pemrogram harus menelusuri lagi baris-per-baris kode-sumber untuk menentukan letak sumber kesalahan ini dan memperbaikinya.

Seorang pemrogram telah menulis kode-sumber seperti di bawah ini:
```

B = 2

```

<span style="color:red">&#9881;</span> Pada sel di bawah ini, tulislah semua sumber kesalahan dari program empat baris di atas dan kemudian perbaikilah sehingga program tersebut dapat berjalan dengan benar.

````python
# Kesalahan pemrograman : Kesalahan pada kode pemrograman tersebut terjadi karena perintah "C = A + B"
#                         diletakkan di awal. Padahal, baik "A" maupun "B" sama-sama belum diketahui nilainya.

# Perbaikan kesalahan pemrograman : Urutan kode program diubah seperti di bawah ini. Urutan diubah supaya
#                                   nilai A dan B diketahui terlebih dahulu sehingga C bisa dihitung oleh Python.

# Jalankan program yang sudah diperbaiki
A = 3
B = 2
C = A + B
print(C)
````

### <span style="color:blue">Tugas #2: Massa dari Berbagai Bahan &#8594; 20 poin</span>

Massa dari berbagai bahan dapat ditentukan berdasarkan rapat massanya. Rapat massa $\rho$ dari suatu bahan didefinisikan dengan rumus:

$$ \rho = \frac{m}{V}  $$

dengan $m$ adalah massa bahan tersebut yang terdapat di dalam suatu volume sebesar $V$.

<span style="color:red">&#9881;</span> Pada sel di bawah ini, tulislah program Python __lengkap__ bernama `MassaBahan.py` yang menghitung dan menampilkan secara rapi massa (gram) dalam satu liter dari bahan-bahan berikut: besi, udara, minyak, es, tubuh manusia, perak dan platinum.

<b>Petunjuk</b>: Gunakan data rapat massa (g/cm$^3$) dari berbagai bahan yang tersimpan di dalam berkas bernama `rapat_massa.dta` yang terdapat di dalam direktori/<i>folder</i> yang sama dengan berkas `LapPPK-02-888888.ipynb` ini. Berkas `rapat_massa.dta` berformat teks sehingga dapat dibaca dengan memakai <i>Jupyter Notebook</i>.

````python
# MassaBahan.py: Program yang menghitung dan menampilkan secara rapi massa (gram) dalam satu liter 
#               dari bahan-bahan seperti besi, udara, minyak, es, tubuh manusia, perak dan platinum.  

# Diketahui dari berkas rapat_massa.dta (dalam g/cm^3)
# rho_minyak = 0.67
# rho_es = 0.9
# rho_tubuh_manusia = 1.03
# rho_besi = 7.8
# rho_perak = 10.5
# rho_platinum = 21.4

# Konversi g/cm^3 ke g/dm^3 atau g/L
rho_minyak = 0.67 * 10**3         # Rapat massa minyak
rho_es = 0.9 * 10**3              # Rapat massa es
rho_tubuh_manusia = 1.03 * 10**3  # Rapat massa tubuh manusia
rho_besi = 7.8 * 10**3            # Rapat massa besi
rho_perak = 10.5 * 10**3          # Rapat massa perak
rho_platinum = 21.4 * 10**3       # Rapat massa platinum

# Rumus
# rho = m / V
# maka,
# m = rho * V

# Diketahui pula volume
V = 1                             # Volume (1 liter)

# Konversi rumus
m_minyak = rho_minyak * V                 # Massa minyak
m_es = rho_es * V                         # Massa es
m_tubuh_manusia = rho_tubuh_manusia * V   # Massa tubuh manusia
m_besi = rho_besi * V                     # Massa besi
m_perak = rho_perak * V                   # Massa perak
m_platinum = rho_platinum * V             # Massa platinum

# Tampilkan program
print('Massa minyak =', m_minyak)
print('Massa es =', m_es)
print('Massa tubuh manusia =', m_tubuh_manusia)
print('Massa besi =', m_besi)
print('Massa perak =', m_perak)
print('Massa platinum =', m_platinum)
````

### <span style="color:blue">Tugas #3: Gaya Hambat &#8594; 20 poin</span>

Gaya hambat atau <i>drag force</i>, yang diakibatkan adanya halangan udara, pada suatu benda dapat diyatakan sebagai rumus berikut:

$$ F_d = \frac{1}{2} \, C_d \, \rho \, A \, v^2 $$

dengan

- $C_d$ = koefisien hambat (<i>drag coefficient</i>) yang nilainya sangat tergantung
pada bentuk benda dan kekasaran permukaan benda;
- $\rho$ = kerapatan udara = 1,2 kg/m$^3$;
- $A$ = luas penampang (tegak lurus terhadap arah gerakan) benda; dan
- $v$ = kecepatan benda.

Rumus di atas dapat dipakai untuk menghitung gaya hambat yang dialami suatu bola kaki ketika ditendang.

<span style="color:red">&#9881;</span> Pada sel di bawah ini, tulislah program Python <b>lengkap</b> bernama `GayaHambat.py` yang menghitung gaya hambat saat suatu bola ditendang. Untuk suatu bola kaki berjejari 11 cm dengan massa = 0,43 kg, koefisien hambat yang terkait adalah 0,4. Tampilkan gaya hambat pada bola tersebut ketika ditendang keras ($v$ = 120 km/jam) dan ketika ditendang lunak ($v$ = 30 km/jam).

````python
# GayaHambat.py : Program yang menghitung gaya hambat saat suatu bola ditendang.

# Diketahui
# r = 11 cm
# v_1 = 120 km/jam
# v_2 = 30 km/jam
m = 0.43                     # Massa bola (kg)
C_d = 0.4                    # Koefisien hambat
rho = 1.2                    # Kerapatan udara (kg/m^3)

# Konversi r menjadi satuan m
r = 11 * 10**-2              # Jari-jari bola (m)

# Konversi v_1 dan v_2 menjadi satuan m/s
v_1 = 120 * 10**3 / 3600     # Kecepatan benda keras
v_2 = 30 * 10**3 / 3600      # Kecepatan benda lambat

# Mencari A (luas penampang bola)
from math import pi
A = 4 * pi * r**2            # Luas penampang bola

# Masukkan rumus gaya hambat saat suatu bola ditendang
F_d1 = 0.5 * C_d * rho * A * v_1**2    # Gaya hambat pada bola ketika ditendang dengan keras
F_d2 = 0.5 * C_d * rho * A * v_2**2    # Gaya hambat pada bola ketika ditendang dengan lunak

# Jalankan program
print('Gaya hambat pada bola ketika ditendang dengan keras =', F_d1)
print('Gaya hambat pada bola ketika ditendang dengan lunak =', F_d2)
````

### <span style="color:blue">Tugas #4: <i>String</i> Gaya Gravitasi &#8594; 20 poin</span>

Gaya gravitasi pada suatu benda dapat dinyatakan dengan rumus:

$$ F_g = m \, g $$

dengan

- $m$ = massa benda; dan
- $g$ = percepatan gravitasi = 9,81 m/s$^2$.

Mirip <font style="color:blue">Tugas #3</font>, rumus di atas dapat dipakai untuk menghitung gaya gravitasi yang dialami suatu bola kaki ketika ditendang.

<span style="color:red">&#9881;</span> Pada sel di bawah ini, tulislah program Python <b>lengkap</b> bernama `GayaGravitasi.py` yang menghitung gaya gravitasi yang dialami suatu bola saat ditendang. Bola yang ditinjau adalah bola yang sama pada <font style="color:blue">Tugas #3</font>. Selain itu, program ini juga menghitung rasio antara gaya hambat dan gaya gravitasi pada bola tersebut ketika ditendang keras ($v$ = 120 km/jam) dan ketika ditendang lunak ($v$ = 30 km/jam).

````python
# Gaya Gravitasi.py : Program yang menghitung gaya gravitasi yang dialami suatu bola ketika ditendang

# Diketahui di Tugas #3
m = 0.43                     # Massa bola (kg)
g = 9.81                     # Percepatan gravitasi (m/s^2)
F_d1 = 40.54748918233227     # Gaya hambat pada bola ketika ditendang dengan keras
F_d2 = 2.534218073895767     # Gaya hambat pada bola ketika ditendang dengan lunak

# Rumus gaya gravitasi
F_g = m * g                  # Gaya gravitasi 

# Rasio antara gaya hambat dan gaya gravitasi 
Rat_1 = F_d1 / F_g           # Rasio antara gaya hambat dan gaya gravitasi (saat ditendang keras)
Rat_2 = F_d2 / F_g           # Rasio antara gaya hambat dan gaya gravitasi (saat ditendang lunak)

# Jalankan program
print('Gaya gravitasi yang dialami suatu bola ketika ditendang =', F_g)
print('Rasio antara gaya hambat dan gaya gravitasi (saat ditendang keras) =', Rat_1)
print('Rasio antara gaya hambat dan gaya gravitasi (saat ditendang lunak) =', Rat_2)
````
